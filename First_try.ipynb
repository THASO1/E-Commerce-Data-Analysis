```python
class BaseDataHandler:
    def __init__(self, file_path):
        self.file_path = file_path
        self.df = None
        self.load_data()
```

```python
class DataLoader(BaseDataHandler):

    def load_data(self):
        try:
            print(f"Loading: {self.file_path}")

            self.df = pd.read_csv(self.file_path)

            print("Data loaded successfully! ✅")

            print("\n----- Dataset Info -----")
            self.df.info()

            print("\n----- Dataset Shape -----")
            print(f"Rows: {self.df.shape[0]:,}, Columns: {self.df.shape[1]}")

            return self.df

        except FileNotFoundError:
            print(f"Error: File '{self.file_path}' was not found.")

        except Exception as e:
            print(f"Unexpected error: {e}")
```

```python
class DataCleaner:

    def __init__(self, data):
        self.df = data.df

    def convert_dates(self, columns=None):
        if columns is None:
            columns = [
                col for col in self.df.columns
                if any(keyword in col.lower()
                       for keyword in ["date", "time", "timestamp", "created_at", "updated_at"])
            ]

        for column in columns:
            self.df[column] = pd.to_datetime(
                self.df[column],
                errors="coerce"
            )

    def remove_duplicates(self):
        self.df.drop_duplicates(inplace=True)

    def fill_missing(self, column, value):
        self.df[column] = self.df[column].fillna(value)

    def fill_missing_median(self, column):
        self.df[column] = self.df[column].fillna(
            self.df[column].median()
        )

    def strip_strings(self):
        string_columns = self.df.select_dtypes(
            include="object"
        ).columns

        for column in string_columns:
            self.df[column] = self.df[column].str.strip()

    def get_dataframe(self):
        return self.df
```

```python
orders_cleaner = DataCleaner(orders)
order_items_cleaner = DataCleaner(order_items)
order_item_refunds_cleaner = DataCleaner(order_item_refunds)
products_cleaner = DataCleaner(products)
pageviews_cleaner = DataCleaner(Web_pageviews)
sessions_cleaner = DataCleaner(Web_sessions)
```

```python
# Convert date columns
orders_cleaner.convert_dates()
order_items_cleaner.convert_dates()
order_item_refunds_cleaner.convert_dates()
products_cleaner.convert_dates()
pageviews_cleaner.convert_dates()
sessions_cleaner.convert_dates()
```

```python
# Orders - Key Performance Indicators

total_orders = orders.df["order_id"].nunique()
total_revenue = orders.df["price_usd"].sum()
total_cogs = orders.df["cogs_usd"].sum()
total_profit = total_revenue - total_cogs
average_order_value = total_revenue / total_orders
profit_margin = (total_profit / total_revenue) * 100

print(f"Total Orders: {total_orders:,}")
print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Total COGS: ${total_cogs:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Average Order Value: ${average_order_value:,.2f}")
print(f"Profit Margin: {profit_margin:.2f}%")
```

```python
# Orders - Monthly Sales Trend

monthly_sales = (
    orders.df
    .set_index("created_at")
    .resample("ME")["price_usd"]
    .sum()
)

monthly_sales
```

```python
monthly_sales.plot(
    kind="line",
    figsize=(12, 5),
    linewidth=2
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (USD)")
plt.grid(alpha=0.3)
plt.show()
```
