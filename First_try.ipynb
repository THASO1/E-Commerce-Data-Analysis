```python
class BaseDataHandler:
    def __init__(self, file_path):
        self.file_path = file_path
        self.df = None
        self.load_data()
```

```python
class DataLoader(BaseDataHandler):

    def load_data(self):
        try:
            print(f"Loading: {self.file_path}")

            self.df = pd.read_csv(self.file_path)

            print("Data loaded successfully! ✅")

            print("\n----- Dataset Info -----")
            self.df.info()

            print("\n----- Dataset Shape -----")
            print(f"Rows: {self.df.shape[0]:,}, Columns: {self.df.shape[1]}")

            return self.df

        except FileNotFoundError:
            print(f"Error: File '{self.file_path}' was not found.")

        except Exception as e:
            print(f"Unexpected error: {e}")
```

```python
class DataCleaner:

    def __init__(self, data):
        self.df = data.df

    def convert_dates(self, columns=None):
        if columns is None:
            columns = [
                col for col in self.df.columns
                if any(keyword in col.lower()
                       for keyword in ["date", "time", "timestamp", "created_at", "updated_at"])
            ]

        for column in columns:
            self.df[column] = pd.to_datetime(
                self.df[column],
                errors="coerce"
            )

    def remove_duplicates(self):
        self.df.drop_duplicates(inplace=True)

    def fill_missing(self, column, value):
        self.df[column] = self.df[column].fillna(value)

    def fill_missing_median(self, column):
        self.df[column] = self.df[column].fillna(
            self.df[column].median()
        )

    def strip_strings(self):
        string_columns = self.df.select_dtypes(
            include="object"
        ).columns

        for column in string_columns:
            self.df[column] = self.df[column].str.strip()

    def get_dataframe(self):
        return self.df
```

```python
refund_analysis = order_item_refunds.df.merge(
    order_items.df[["order_item_id", "price_usd"]],
    on="order_item_id",
    how="left"
)

total_refunds = refund_analysis["order_item_id"].nunique()
refunded_revenue = refund_analysis["price_usd"].sum()
refund_rate = total_refunds / order_items.df["order_item_id"].nunique() * 100

print(f"Total Refunded Items: {total_refunds:,}")
print(f"Refunded Revenue: ${refunded_revenue:,.2f}")
print(f"Refund Rate: {refund_rate:.2f}%")
```

### Key Insights
- 1,731 items were refunded, representing a 4.32% refund rate.
- Refunded items represent approximately $85.3K in revenue.
- The overall refund rate is relatively low, indicating that most purchases are retained.
- Product-level refund analysis can reveal which products contribute most to refunds and may require further investigation.

```python
product_refunds = (
    order_items.df
    .groupby("product_id")
    .agg(
        Total_Items=("order_item_id", "count")
    )
    .reset_index()
)

refunded_items = (
    order_item_refunds.df
    .groupby("order_item_id")
    .size()
    .reset_index(name="Refunded")
)

product_refunds = (
    order_items.df[["order_item_id", "product_id"]]
    .merge(refunded_items, on="order_item_id", how="left")
)

product_refunds["Refunded"] = product_refunds["Refunded"].fillna(0)

product_refunds = (
    product_refunds
    .groupby("product_id")
    .agg(
        Total_Items=("order_item_id", "count"),
        Refunded_Items=("Refunded", lambda x: (x > 0).sum())
    )
    .reset_index()
)

product_refunds["Refund_Rate"] = (
    product_refunds["Refunded_Items"] /
    product_refunds["Total_Items"] * 100
)

product_refunds = product_refunds.merge(
    products.df[["product_id", "product_name"]],
    on="product_id",
    how="left"
).sort_values("Refund_Rate", ascending=False)

product_refunds
```

```python
product_refunds.plot(
    x="product_name",
    y="Refund_Rate",
    kind="bar",
    figsize=(10, 5),
    legend=False
)

plt.title("Refund Rate by Product")
plt.xlabel("Product")
plt.ylabel("Refund Rate (%)")
plt.xticks(rotation=20)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

### Key Insights
- The Birthday Sugar Panda has the highest refund rate at **6.04%**, indicating a potential product-level issue.
- The Original Mr. Fuzzy has the second-highest refund rate at **5.11%** and also accounts for the largest number of refunded items.
- The Forever Love Bear has a lower refund rate of **2.23%**, while the Hudson River Mini Bear has the lowest at **1.28%**.
- Refund rates vary significantly across products, suggesting that product-level factors may influence customer returns.
- The Birthday Sugar Panda should be prioritized for further investigation despite having a lower sales volume than Mr. Fuzzy.

```python
monthly_business = (
    Web_sessions.df
    .assign(month=Web_sessions.df["created_at"].dt.to_period("M"))
    .groupby("month")
    .agg(
        Sessions=("website_session_id", "nunique")
    )
    .reset_index()
)

monthly_orders_revenue = (
    orders.df
    .assign(month=orders.df["created_at"].dt.to_period("M"))
    .groupby("month")
    .agg(
        Orders=("order_id", "nunique"),
        Revenue=("price_usd", "sum")
    )
    .reset_index()
)

monthly_business = monthly_business.merge(
    monthly_orders_revenue,
    on="month",
    how="left"
)

monthly_business["Orders"] = monthly_business["Orders"].fillna(0)
monthly_business["Revenue"] = monthly_business["Revenue"].fillna(0)

monthly_business["Conversion_Rate"] = (
    monthly_business["Orders"] /
    monthly_business["Sessions"] * 100
)

monthly_business["month"] = monthly_business["month"].astype(str)

monthly_business
```

### Key Insights
- Monthly revenue increased substantially as both website traffic and conversion rate improved.
- Revenue reached its highest point in **December 2014 at $144.8K**, supported by 29,722 sessions and a 7.79% conversion rate.
- The highest conversion rate occurred in **February 2015 at 8.70%**, but revenue was lower at $129.2K because traffic was lower than in December 2014.
- This shows that **conversion efficiency alone does not determine revenue; traffic volume also plays a major role**.
- The business achieved stronger revenue growth through a combination of increasing traffic and improving website conversion.
- Average order value is effectively stable at about **$49.99**, so changes in revenue are primarily driven by the number of orders rather than changes in order value.

```python
customer_orders = (
    orders.df
    .sort_values("created_at")
    .assign(
        Order_Number=lambda x: x.groupby("user_id").cumcount() + 1,
        month=lambda x: x["created_at"].dt.to_period("M")
    )
)

monthly_retention = (
    customer_orders
    .groupby("month")
    .agg(
        Total_Orders=("order_id", "nunique"),
        Repeat_Orders=("Order_Number", lambda x: (x > 1).sum())
    )
    .reset_index()
)

monthly_retention["Repeat_Order_Rate"] = (
    monthly_retention["Repeat_Orders"] /
    monthly_retention["Total_Orders"] * 100
)

monthly_retention["month"] = monthly_retention["month"].astype(str)

monthly_retention
```

```python
monthly_retention.plot(
    x="month",
    y="Repeat_Order_Rate",
    kind="line",
    figsize=(12, 5),
    linewidth=2
)

plt.title("Monthly Repeat Order Rate")
plt.xlabel("Month")
plt.ylabel("Repeat Order Rate (%)")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.show()
```

### Key Insights
- Repeat orders represented less than 1% of monthly orders during the early stages of the business.
- The repeat order rate gradually improved throughout 2013 and 2014.
- The rate reached **3.19% in March 2015**, the highest level in the analyzed period.
- Despite this improvement, repeat orders still represented only a small share of total orders.
- The increasing repeat order rate suggests improving customer retention, but the business remains heavily dependent on acquiring new customers.

```python
customer_segments = (
    orders.df
    .groupby("user_id")
    .agg(
        Orders=("order_id", "nunique"),
        Revenue=("price_usd", "sum")
    )
    .reset_index()
)

customer_segments["Segment"] = pd.cut(
    customer_segments["Orders"],
    bins=[0, 1, 2, 3],
    labels=["One-Time", "Repeat", "High-Frequency"]
)

segment_analysis = (
    customer_segments
    .groupby("Segment", observed=False)
    .agg(
        Customers=("user_id", "nunique"),
        Total_Revenue=("Revenue", "sum"),
        Avg_Revenue_Per_Customer=("Revenue", "mean"),
        Avg_Orders=("Orders", "mean")
    )
    .reset_index()
)

segment_analysis["Revenue_Share"] = (
    segment_analysis["Total_Revenue"] /
    segment_analysis["Total_Revenue"].sum() * 100
)

segment_analysis
```

```python
segment_analysis.plot(
    x="Segment",
    y="Avg_Revenue_Per_Customer",
    kind="bar",
    figsize=(9, 5),
    legend=False
)

plt.title("Average Customer Value by Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Average Revenue per Customer (USD)")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

### Key Insights
- One-Time customers generate the majority of total revenue (**96.16%**) because they represent almost the entire customer base.
- Repeat customers generate more than twice the average revenue of One-Time customers (**$123.34 vs. $59.93**).
- High-Frequency customers have the highest average customer value at **$179.58**, approximately three times the value of a One-Time customer.
- Only **565 Repeat** and **26 High-Frequency** customers exist, highlighting a significant customer retention opportunity.
- Increasing the number of customers who make repeat purchases could substantially increase customer lifetime value and reduce dependence on continuous customer acquisition.
